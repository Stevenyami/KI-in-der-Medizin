## Aufgabe 3: Datenanalyse (Vorrechenübung)


*Diese Aufgabe ist eine Vorrechenübung: Der gesamte Code ist bereits vorgegeben, ihr müsst nichts mehr selbst implementieren. Lest den Code und die Erklärungen trotzdem aufmerksam durch, das Vorgehen kann in ähnlicher Form in der Klausur gefragt werden.*

Ein sehr wichtiger Schritt beim Machine Learning ist die genaue Analyse des Datensatzes - das gilt besonders, wenn ihr ihn nicht selbst erhoben habt und die Spalten nicht aus erster Hand kennt. Wir schauen uns das am Beispiel der Spalte *time* an: Sie gibt an, wie viele Tage eine Person im Beobachtungszeitraum der Studie nachverfolgt wurde, bevor sie entweder verstarb (`DEATH_EVENT=1`) oder die Studie endete (`DEATH_EVENT=0`).

Zunächst stellen wir den Zusammenhang zwischen *time* und *DEATH_EVENT* grafisch dar.

```python
import matplotlib.pyplot as plt

# Streudiagramm: jeder Punkt ist eine Person
plt.scatter(imported_dataset['time'], imported_dataset['DEATH_EVENT'])
plt.xlabel('time (Tage)')
plt.ylabel('DEATH_EVENT')
plt.show()
```

Da `DEATH_EVENT` nur die Werte 0 und 1 annimmt, ist ein Streudiagramm hier nicht ideal - alle Punkte liegen auf zwei horizontalen Linien. Aussagekräftiger ist ein Histogramm, das die Verteilung von *time* für beide Gruppen (gestorben / nicht gestorben) getrennt zeigt.

```python
plt.hist(imported_dataset['time'][imported_dataset['DEATH_EVENT'] == 0], bins=20)
plt.hist(imported_dataset['time'][imported_dataset['DEATH_EVENT'] == 1], bins=20)
plt.legend(['No Death', 'Death'])
plt.xlabel('time (Tage)')
plt.ylabel('Anzahl Personen')
plt.show()
```

Man erkennt: Personen, die gestorben sind (orange), haben fast ausschließlich einen *kleinen* Wert von *time*, während Personen, die überlebt haben (blau), über einen viel größeren Bereich verteilt sind, oft mit deutlich größerem *time*. Das bestätigen wir jetzt noch mit der Korrelation.

```python
cor_time_death = imported_dataset[['time', 'DEATH_EVENT']].corr()
print("Korrelation zwischen den Spalten 'time' und 'DEATH_EVENT' beträgt:\n", cor_time_death)
```

Die Korrelation ist stark negativ: je kleiner *time*, desto wahrscheinlicher `DEATH_EVENT=1`. Das liegt daran, wie *time* erhoben wurde: Für Personen, die während der Studie gestorben sind, endet die Nachverfolgung logischerweise mit ihrem Tod - ihr *time*-Wert ist also quasi die "Überlebenszeit nach Studienbeginn". Für alle anderen läuft *time* bis zum Studienende weiter.

Das Problem: Zum Zeitpunkt, an dem wir eine Vorhersage treffen wollen (z.B. bei Aufnahme der Person), wissen wir noch gar nicht, wie lange sie nachbeobachtet werden wird - dieser Wert existiert in der Realität erst im Nachhinein. Ein Modell, das *time* als Feature nutzt, hätte also Zugriff auf Information, die es zum Vorhersagezeitpunkt in der Praxis nie haben könnte. Dieses Phänomen nennt man **Data Leakage**: Eine im Datensatz vorhandene Spalte "verrät" quasi direkt oder indirekt das Label, obwohl sie im späteren Einsatz nicht zur Verfügung steht.

Bevor wir die Spalte entfernen, schauen wir uns noch an, wie man sich systematisch (statt nur mit einzelnen Plots) einen Überblick über einen Datensatz verschafft. Pandas stellt dafür mehrere nützliche Befehle bereit:
* `dataframe.shape` gibt die Anzahl der Zeilen (Personen) und Spalten (Features) zurück.
* `dataframe.dtypes` zeigt den von Pandas erkannten Datentyp jeder Spalte (z.B. `int64`, `float64`).
* `dataframe.describe()` berechnet für jede numerische Spalte Kennzahlen wie Mittelwert, Minimum, Maximum und Quartile.
* `dataframe.isnull().sum()` zählt pro Spalte, wie viele fehlende (`NULL`) Werte vorhanden sind.

```python
imported_dataset.shape
```

```python
imported_dataset.dtypes
```

Wichtig: `dtypes` ist nicht immer aussagekräftig! Die Spalten `anaemia`, `diabetes`, `sex`, `smoking` und `DEATH_EVENT` sind zwar als Zahlen (0/1) kodiert und erscheinen deshalb als numerischer Typ, inhaltlich handelt es sich aber um nominal skalierte kategoriale Daten (siehe Aufgabe 1a) - alle anderen Spalten sind tatsächlich numerisch.

Mit `isnull()` prüfen wir, ob es fehlende Werte gibt, die wir vor einer Modellierung behandeln müssten (z.B. durch Imputation, siehe die Klausuraufgabe aus Übung 9 bzw. die dort besprochene Gefahr von Data Leakage bei der Imputation).
